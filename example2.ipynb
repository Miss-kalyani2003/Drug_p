```python
import numpy as np
import pandas as pd


from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import json 
import pickle
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("drug.csv")
df.head()
```

```python
df['Sex'].value_counts().to_dict()
```

```python
df['BP'].value_counts().to_dict()
```

```python
df['Cholesterol'].value_counts().to_dict()
```

```python
df['Sex'].replace({'M': 1, 'F': 0},inplace=True)
```

```python
df['BP'].replace({'HIGH': 2, 'LOW': 0, 'NORMAL': 1},inplace=True)
```

```python
df['Cholesterol'].replace({'HIGH': 1, 'NORMAL': 0},inplace=True)
```

```python
df.info()
```

```python
x=df.drop('Drug',axis=1)
y=df['Drug']
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print("Accuracy",accuracy_score(y_test,y_pred))
```

```python
with open("logisticRegression.pkl",'wb') as f:
    pickle.dump(model,f)
    
```
